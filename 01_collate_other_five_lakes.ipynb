```python
import pandas as pd
import xarray as xr
import numpy as np
import os

OUTPUT_DIR = "Code Outputs/Lake Level Outputs"
os.makedirs(OUTPUT_DIR, exist_ok=True)
def out(n): return os.path.join(OUTPUT_DIR, n)

# Define extraction functions
def load_grealm_data(filepath, col_name):
    if not os.path.exists(filepath):
        print(f"  [!] G-REALM file not found: {filepath}")
        return pd.DataFrame()
        
    data = []
    with open(filepath, 'r') as f:
        for line in f:
            if line.startswith(('TOPEX', 'JASN1', 'JASN2', 'JASN3', 'SEN6A', 'POSDN', 'S3A')):
                parts = line.split()
                if len(parts) >= 15:
                    date_str = parts[2]     
                    level_str = parts[14]   
                    if date_str != '99999999' and level_str != '9999.99':
                        data.append({
                            'Date': pd.to_datetime(date_str, format='%Y%m%d').normalize(), 
                            col_name: float(level_str)
                        })
    if not data:
        return pd.DataFrame()
        
    df = pd.DataFrame(data).set_index('Date')
    return df.resample('D').mean()

def load_copernicus_data(filepath, col_name):
    if not os.path.exists(filepath):
        print(f"  [!] Copernicus file not found: {filepath}")
        return pd.DataFrame()
        
    try:
        dataset = xr.open_dataset(filepath)
        df_nc = dataset.to_dataframe().reset_index()
        
        time_col = 'time' if 'time' in df_nc.columns else [c for c in df_nc.columns if 'time' in c.lower()][0]
        df_nc[time_col] = pd.to_datetime(df_nc[time_col]).dt.normalize()
        df_nc.set_index(time_col, inplace=True)

        if 'water_surface_height_above_reference_datum' in df_nc.columns:
            target_col = 'water_surface_height_above_reference_datum'
        elif 'water_surface_height' in df_nc.columns:
            target_col = 'water_surface_height'
        else:
            target_col = [c for c in df_nc.columns if 'height' in c.lower()][0]

        df_daily = df_nc[[target_col]].resample('D').mean()
        df_daily.rename(columns={target_col: col_name}, inplace=True)
        df_daily.index.name = 'Date'
        return df_daily
    except Exception as e:
        print(f"  [!] Error loading {filepath}: {e}")
        return pd.DataFrame()

# Setup the Master File
master_file = out('African_Great_Lakes_Water_Levels.xlsx')
print(f"1. Loading master file: {master_file}...")
df_master = pd.read_excel(master_file)

df_master['Date'] = pd.to_datetime(df_master['Date']).dt.normalize()
df_master['Reservoir'] = df_master['Reservoir'].str.strip()
df_master = df_master[df_master['Date'].dt.year > 1990] # Kill the 1970 rows
df_master.set_index(['Date', 'Reservoir'], inplace=True)

# Process the 5 new lakes
lakes = ['Malawi', 'Albert', 'Edward', 'Kivu', 'Turkana']
dfs_to_combine = []

print("\n2. Extracting data for the 5 lakes...")
for lake in lakes:
    full_lake_name = f'Lake {lake}'
    print(f"\nProcessing {full_lake_name}...")
    lake_data_sources = []
    
    # Get G-REALM
    grealm_path = f"Grealm Data/{lake} Water Level.txt"
    df_g = load_grealm_data(grealm_path, 'GREALM_Level_m')
    if not df_g.empty:
        lake_data_sources.append(df_g)
        print("  -> G-REALM loaded")

    # Get Copernicus
    cop_path = f"Copernicus Data/Water Level Satellite Data/{lake}_Data.nc"
    df_c = load_copernicus_data(cop_path, 'Copernicus_Level_m')
    if not df_c.empty:
        lake_data_sources.append(df_c)
        print("  -> Copernicus loaded")

    # Get DAHITI
    dahiti_path = out(f"{lake}_DAHITI.xlsx")
    if os.path.exists(dahiti_path):
        df_d = pd.read_excel(dahiti_path)
        df_d['Date'] = pd.to_datetime(df_d['Date']).dt.normalize()
        df_d.set_index('Date', inplace=True)
        if 'DAHITI_Level_m' in df_d.columns:
            # Only want the Date index and the level column
            lake_data_sources.append(df_d[['DAHITI_Level_m']])
            print("  -> DAHITI loaded")

    # Combine all available sources for this specific lake
    if lake_data_sources:
        # Start with the first source, join the others safely on the Date index
        df_lake_merged = lake_data_sources[0]
        for i in range(1, len(lake_data_sources)):
            df_lake_merged = df_lake_merged.join(lake_data_sources[i], how='outer')
        
        # Remove any rogue 1970 dates generated by NetCDF resamples
        df_lake_merged = df_lake_merged[df_lake_merged.index.year > 1990]
        
        # Assign the Reservoir name
        df_lake_merged['Reservoir'] = full_lake_name
        df_lake_merged.set_index('Reservoir', append=True, inplace=True)
        
        dfs_to_combine.append(df_lake_merged)

# Merge Everything Together
if dfs_to_combine:
    print("\n3. Merging all new data into the master file...")
    df_new_all = pd.concat(dfs_to_combine)
    
    # Use combine_first to safely put the new data into the master file
    df_master = df_master.combine_first(df_new_all)
    df_master.reset_index(inplace=True)
    
    # Final Cleanup
    sat_cols = ['DAHITI_Level_m', 'GREALM_Level_m', 'Copernicus_Level_m']
    existing_cols = [col for col in sat_cols if col in df_master.columns]
    
    # Drop rows where ALL satellites are completely blank
    rows_before = len(df_master)
    df_master.dropna(subset=existing_cols, how='all', inplace=True)
    print(f"  -> Deleted {rows_before - len(df_master)} completely empty rows.")

    # Calculate overall average
    df_master['Average_Level_m'] = df_master[existing_cols].mean(axis=1)
    
    # Sort neatly by Lake, then by Date
    df_master.sort_values(by=['Reservoir', 'Date'], inplace=True)
    
    # Backup and Save
    df_master.to_excel(out('African_Great_Lakes_Water_Levels_BACKUP.xlsx'), index=False)
    df_master.to_excel(master_file, index=False)
    print(f"\nAll 7 lakes are now collated in {master_file}.")
else:
    print("\nNo new data was found to merge.")
```
